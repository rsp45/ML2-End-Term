```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN
```

```python
# Generate non-spherical data (where K-Means would fail)
X, _ = make_moons(n_samples=300, noise=0.08, random_state=42)
```

```python
X
```

```python
# Hyperparameters: eps (radius) and min_samples (density threshold)
# Try changing eps to 0.05 (too small, everything becomes noise) or 0.3 (too big, merges clusters)
dbscan = DBSCAN(eps=0.15, min_samples=6)
labels = dbscan.fit_predict(X)
```

```python
# Separate points for visualization
core_samples_mask = np.zeros_like(dbscan.labels_, dtype=bool)
core_samples_mask[dbscan.core_sample_indices_] = True

unique_labels = set(labels)
colors = [plt.cm.Spectral(each) for each in np.linspace(0, 1, len(unique_labels))]
```

```python
plt.figure(figsize=(10, 6))
for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = [0, 0, 0, 1]
        label_name = 'Noise (Outliers)'
    else:
        label_name = f'Cluster {k}'

    class_member_mask = (labels == k)
    
    # Plot Core Points (Large)
    xy = X[class_member_mask & core_samples_mask]
    plt.plot(xy[:, 0], xy[:, 1], 'o', markerfacecolor=tuple(col), markeredgecolor='k', markersize=10, label=label_name if 'Core' not in label_name else "")

    # Plot Border Points (Small)
    xy = X[class_member_mask & ~core_samples_mask]
    plt.plot(xy[:, 0], xy[:, 1], 'o', markerfacecolor=tuple(col), markeredgecolor='k', markersize=4)

plt.title(f"DBSCAN Clustering (eps={dbscan.eps}, min_samples={dbscan.min_samples})\nLarge dots = Core, Small dots = Border, Black = Noise")
plt.legend()
plt.show()
```

`eps` ($\epsilon$): The maximum distance between two samples for one to be considered as in the neighborhood of the other.

`min_samples`: The number of samples (or total weight) in a neighborhood for a point to be considered as a core point.
